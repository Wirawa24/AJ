```python
from datasets import load_dataset
import pandas as pd
```

```python
dataset = load_dataset("ReactiveAI/passkey-retrieval", "to-4k")
```

```python
import re
import pandas as pd

# Create a DataFrame from our dataset if not already done
df_train = pd.DataFrame(dataset['train'])

passkeys = set()

# Standard passkey-retrieval pattern: look for names of passkeys in queries/answers
# Usually they appear in contexts like: "The passkey is 'XXXX'" or "private passkey is 'XXXX'"
for index, row in df_train.iterrows():
    for item in row['interactions']:
        query = item.get('query', '')
        answer = item.get('answer', '')

        # Search in both query and answer for phrases like "passkey is '...'"
        for text in [query, answer]:
            matches = re.findall(r"(?:passkey|password|code)\s+is\s+['\"]([^'\"]+?)['\"]", text, re.IGNORECASE)
            for match in matches:
                cleaned = match.strip()
                # Simple heuristic to make sure it's a passkey (not a massive paragraph)
                if 2 < len(cleaned) < 40:
                    passkeys.add(cleaned)

print(f"Total unique passkeys found: {len(passkeys)}")
print("Sample of unique passkeys:", sorted(list(passkeys))[:10])
```
